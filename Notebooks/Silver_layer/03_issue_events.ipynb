## Create the `issue_events` Silver Table

This step creates the Delta table used to store cleaned and structured **IssuesEvent** records in the Silver layer.

The table contains three groups of fields:

- **Common event fields** — General GitHub event information such as the event ID, timestamp, actor, and repository.
- **Bronze lineage fields** — Metadata retained from the Bronze layer for traceability back to the original source data.
- **IssuesEvent-specific fields** — Information describing the issue lifecycle, including the action performed, issue ID, issue number, creation timestamp, and closing timestamp.

These fields allow the Silver layer to track how issues are created, updated, closed, reopened, and otherwise changed over time.

The table is stored using **Delta Lake** (`USING DELTA`).

`CREATE TABLE IF NOT EXISTS` allows the notebook to be rerun safely without replacing an already existing table.

> **Note:** This step defines only the Silver table structure. The extraction, transformation, cleaning, and validation of IssuesEvent data are handled in subsequent steps.

```python
%sql
CREATE TABLE IF NOT EXISTS silver.issue_events (
    -- Common fields
    event_id         STRING,
    event_type       STRING,
    event_timestamp  TIMESTAMP,
    event_date       DATE,
    actor_id         BIGINT,
    actor_login      STRING,
    repo_id          BIGINT,
    repo_name        STRING,

    -- Event-specific fields
    issue_action      STRING,
    issue_id          BIGINT,
    issue_number      BIGINT,
    issue_created_at  TIMESTAMP,
    issue_closed_at   TIMESTAMP,

    -- Bronze lineage
    _source_file     STRING,
    _source_date     DATE,
    _ingested_at     TIMESTAMP
)
USING DELTA;
```
