## Create the `watch_events` Silver Table

This step creates the Delta table used to store cleaned and structured **WatchEvent** records in the Silver layer.

In GitHub event data, a `WatchEvent` primarily represents **repository star activity**.

The table contains three groups of fields:

- **Common event fields** — General event information such as event ID, timestamp, actor, and repository.
- **Bronze lineage fields** — Metadata retained from Bronze for source traceability.
- **WatchEvent-specific fields** — The action associated with the watch event.

Because WatchEvent has a relatively simple payload, only a small number of event-specific fields need to be retained.

These records can later be used to analyze repository interest, popularity, and star activity over time.

The table is stored using **Delta Lake** (`USING DELTA`).

`CREATE TABLE IF NOT EXISTS` ensures the table is created only when it does not already exist.

> **Note:** This step defines the Silver table structure only. Filtering, cleaning, validation, and loading of WatchEvent records are performed in following steps.

```python
%sql
CREATE TABLE IF NOT EXISTS silver.watch_events (
    -- Common fields
    event_id         STRING,
    event_type       STRING,
    event_timestamp  TIMESTAMP,
    event_date       DATE,
    actor_id         BIGINT,
    actor_login      STRING,
    repo_id          BIGINT,
    repo_name        STRING,

    -- Event-specific fields
    watch_action     STRING,

    -- Bronze lineage
    _source_file     STRING,
    _source_date     DATE,
    _ingested_at     TIMESTAMP
)
USING DELTA;
```
